### Series Maths Methods

### 1. The Definitions
* **`size`**: The **Container Dimensions**. It counts the number of "slots" or rows in your series. It includes everything—valid data, zeros, and even missing values (`NaN`).
* **`count()`**: The **Valid Data Counter**. It counts only the *non-empty* (non-NaN) values. It ignores missing data.
* **`sum()`**: The **Mathematical Total**. It adds up the values.
    * For Numbers: Arithmetic sum ($10 + 20 = 30$).
    * For Strings: Concatenation ("A" + "B" = "AB").

### 2. Applied to Your Datasets

#### Case A: `vk` (Kohli IPL Runs) - Numerical Data
* **Data:** A list of runs scored in 215 matches.
* **`vk.size` -> 215**
    * *Meaning:* There are 215 records/rows in this file.
* **`vk.count()` -> 215**
    * *Meaning:* We have valid run data for all 215 matches (no missing info).
* **`vk.sum()` -> 6634**
    * *Meaning:* The total runs scored across all matches ($1 + 23 + 13 + ...$).

#### Case B: `subs` (Subscribers Gained) - Numerical Data
* **Data:** Daily subscriber gains for 365 days.
* **`subs.size` -> 365**
    * *Meaning:* The dataset covers 365 days.
* **`subs.count()` -> 365**
    * *Meaning:* Every single day has a recorded number (no missing days).
* **`subs.sum()` -> 49,510**
    * *Meaning:* Total subscribers gained in the entire year.

### Boolean Indexing on Series

````python
import numpy as np
vk = np.array([20, 55, 40, 80, 10])
vk
````

### Truth Values (Booleans)
The comparison generates an array of boolean values.

$$
\begin{bmatrix} 20 \\ 55 \\ 40 \\ 80 \\ 10 \end{bmatrix} \ge 50 \quad \longrightarrow \quad \begin{bmatrix} \text{False} \\ \text{True} \\ \text{False} \\ \text{True} \\ \text{False} \end{bmatrix}
$$

This resulting array is called the **Boolean Mask**.

### Step 2: The Outer Operation (Applying the Mask)
Now the code looks like this to the computer: `vk[mask]`.

The outer brackets tell NumPy to "overlay" the mask on top of the original `vk` array.

| Index | Original `vk` | Mask (`vk >= 50`) | Action |
| :--- | :--- | :--- | :--- |
| 0 | 20 | **False** | ❌ Block |
| 1 | 55 | **True** | ✅ **Pass** |
| 2 | 40 | **False** | ❌ Block |
| 3 | 80 | **True** | ✅ **Pass** |
| 4 | 10 | **False** | ❌ Block |

### Final Result
The computer collects only the values that "Passed" the mask and creates a **new** array in memory.

- ## `apply` function like loop

| Method | Logic | Manual Work |
| :--- | :--- | :--- |
| **For Loop** | "Get length of list, start at index 0, get value, change value, save to new list, move to index 1..." | High |
| **Apply** | "Take this rule (add_tax) and run it on this list (prices)." | Low |





- ## Imagine you have a list of prices in a Series, and you want to calculate a 10% tax for each one.

````python
def add_tax(price):
    return price * 1.10

import pandas as pd

# Our data (The Belt)
prices = pd.Series([100, 200, 300, 400])

# The Power Switch (Apply)
taxed_prices = prices.apply(add_tax)

print(taxed_prices)
````
