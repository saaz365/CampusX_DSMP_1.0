- ### 40. **isnull/dropna:** If a Series has `NaN` values, count how many there are and then drop them.

- ###  eg 1,2,3,np.nan,5,6,np.nan,8,np.nan,10


```python
import numpy as np
import pandas as pd

data = pd.Series( [1,2,3,np.nan,5,6,np.nan,8,np.nan,10])
data
```

```python
data.size
```

```python
data.count()
```

```python
data.isnull().count()
```

```python
data.isnull().size
```

```python
data.isnull().sum()
```

```python
data[data.isnull()]
```

```python
data[~data.isnull()]
```

```python
data.dropna()
```

```python
data
```

```python
data.fillna()
```

```python
data.fillna("x")
```

```python
data.fillna(data.mean())
```

```python
data.fillna( 0)
```
