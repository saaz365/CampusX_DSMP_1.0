---
- ### In simple terms, the **`rank()`** method in Pandas is used to assign a numerical position (a rank) to each element in a dataset based on its value relative to other elements in the same column.

While **sorting** rearranges the data, **ranking** assigns a score (1, 2, 3...) to the data without changing its original order.

---

## 1. The Core Arguments

To use `rank()` effectively, you need to understand three main settings:

* **`ascending`**:
* `True` (Default): Smallest value gets Rank 1.
* `False`: Largest value gets Rank 1 (e.g., top scorers).


* **`method`**: Decides how to handle ties (when two values are the same).
* **`pct`**: If set to `True`, it returns the rank as a percentile (e.g., 0.85 instead of Rank 85).

---

## 2. Tie-Breaking Methods (The "How")

This is where most users get confused. If two people have the same score, how do you rank them?

| Method | Description | Result for [10, 20, 20, 30] |
| --- | --- | --- |
| **`average`** | Takes the average of the ranks (Default). | `1, 2.5, 2.5, 4` |
| **`min`** | Both get the lowest rank (like the Olympics). | `1, 2, 2, 4` |
| **`max`** | Both get the highest rank. | `1, 3, 3, 4` |
| **`dense`** | Like `min`, but the next rank is always +1 (no gaps). | `1, 2, 2, 3` |
| **`first`** | Ranks them based on who appears first in the list. | `1, 2, 3, 4` |

---

## 3. Practical Example

Let's say you have a list of players and their runs:

```python
import pandas as pd

df = pd.DataFrame({
    'Player': ['A', 'B', 'C', 'D'],
    'Runs': [100, 80, 80, 50]
})

# Who is the top scorer?
df['Rank'] = df['Runs'].rank(ascending=False, method='min')

print(df)

```

**Output:**
| Player | Runs | Rank |
| :--- | :--- | :--- |
| A | 100 | 1.0 |
| B | 80 | 2.0 |
| C | 80 | 2.0 |
| D | 50 | 4.0 |

> **Note:** Because B and C are tied for 2nd, the next rank is 4.0. There is no Rank 3 because two people occupied the "2nd and 3rd" slots.

---

## 4. Why use Rank instead of Sort?

1. **Preserve Index:** You keep your data in its original order but still know who is "First."
2. **Filtering:** You can easily filter the "Top 10" by doing `df[df['Rank'] <= 10]`.
3. **Statistical Analysis:** Ranking is essential for non-parametric statistics (like Spearman's Correlation).



````python
import pandas as pd

df = pd.DataFrame({
    'Player': ['A', 'B', 'C', 'D'],
    'Runs': [100, 80, 80, 50]
})

# Who is the top scorer?
df['Rank'] = df['Runs'].rank(ascending=False, method='min')

print(df)
````

---
- ### To explain how this works, let's look at the **internal logic** Pandas follows when you run that specific line of code.

### The Step-by-Step Internal Process

When you execute `df['Runs'].rank(ascending=False, method='min')`, Pandas performs these three steps:

**Step 1: Sorting (Internal)**
It looks at the `Runs` column and sorts them based on your `ascending=False` instruction (Highest to Lowest):

1. 100
2. 80
3. 80
4. 50

**Step 2: Assigning Positions**
It assigns "virtual" positions to these values:

* 100 is at Position **1**
* 80 is at Position **2**
* 80 is at Position **3**
* 50 is at Position **4**

**Step 3: Applying the `method='min'` Rule**
Since there is a tie (two 80s at positions 2 and 3), the `min` method tells Pandas: *"If values are equal, give everyone the **minimum** position number among them."*

* The 80s are at positions 2 and 3. The minimum is **2**.
* Both 80s are now **Rank 2**.
* The next value (50) remains at **Rank 4**.

---

### The Output Explained

Your code will produce this DataFrame:

|  | Player | Runs | Rank | Why? |
| --- | --- | --- | --- | --- |
| **0** | A | 100 | **1.0** | Highest score. |
| **1** | B | 80 | **2.0** | Tied for 2nd; gets the minimum rank (2). |
| **2** | C | 80 | **2.0** | Tied for 2nd; gets the minimum rank (2). |
| **3** | D | 50 | **4.0** | 4th person in line (Rank 3 is skipped). |

---

### Comparison: Changing the `method`

To see the difference, look what happens to the same data if we change the method:

| Value | `min` (Olympic) | `dense` (No Gaps) | `average` (Default) |
| --- | --- | --- | --- |
| 100 | 1.0 | 1.0 | 1.0 |
| 80 | **2.0** | **2.0** | **2.5** |
| 80 | **2.0** | **2.0** | **2.5** |
| 50 | **4.0** | **3.0** | **4.0** |

* **`min`**: Great for "Top 3" lists where ties take up the next spot.
* **`dense`**: Best for when you want "Rank 1, 2, 3" without skipping numbers, regardless of ties.
* **`average`**: Used in statistics; it splits the difference between the 2nd and 3rd spot ().

### A quick tip:

Notice that the output is `1.0` (a float). This is because the `average` method often creates decimals, so Pandas defaults the whole column to floats. If you want integers, you can add `.astype(int)` at the end.



````python
import pandas as pd

data = {
    'batter': ['Virat Kohli', 'MS Dhoni', 'Rohit Sharma', 'Suresh Raina'],
    'batsman_run': [6000, 5000, 5000, 5500]
}
df = pd.DataFrame(data)

# Applying Rank
df['rank_desc'] = df['batsman_run'].rank(ascending=False)
print(df)
````

````python
def euclidean(row):
  pt_A = row['1st point']
  pt_B = row['2nd point']

  return ((pt_A[0] - pt_B[0])**2 + (pt_A[1] - pt_B[1])**2)**0.5
````

````python
points_df['distance'] = points_df.apply(euclidean,axis=1)
points_df
````
